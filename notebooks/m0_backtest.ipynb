```python
import tree_sitter_python as tspython
from tree_sitter import Language, Parser

PY_LANGUAGE = Language(tspython.language())
parser = Parser(PY_LANGUAGE)


```

```python
tree = parser.parse(b"import numpy as np")
print(tree.root_node)
```

```python
samples = [
    b"from numpy import float_ as f",
    b"x = np.float_(1)",
    b"import numpy.linalg as la",
    b"from numpy import *",
    b"a.b.c",
]

for sample in samples:
  sample_tree = parser.parse(sample)
  print(sample_tree.root_node)
```

2. expression statment, left is identifier, right calls a function, is an attribute whose object is np and whose attribute is float
3. import without the from as an alias
4. import from, with identifier being a wildcard import
